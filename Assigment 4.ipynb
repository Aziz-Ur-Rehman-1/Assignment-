```python
#Q3 — Loop Through Dictionary
#Print keys and values using loop.

person = {
    "name": "Aziz Ur Rehman",
    "age": 20,
    "city": "Hyderabad"
}

for key, value in person.items():
    print(f"{key}: {value}")
```

```python
#Q4 — Count Frequency of Numbers
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}
for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print(frequency)
```

```python
#Q5 — Word Counter
#Take sentence input and count word frequency.

sentence = input("Enter a sentence: ")
words = sentence.lower().split()
word_frequency = {}

for word in words:
    word = word.strip('.,!?;:()"\'')
    if word in word_frequency:
        word_frequency[word] += 1
    else:
        word_frequency[word] = 1

print("\nWord Frequency:")
for word, count in word_frequency.items():
    print(f"{word}: {count}")
```

```python
#Q6 — Student Marks & Average
marks = {"Math": 80, "English": 70, "Physics": 90}
total = sum(marks.values())
average = total / len(marks)

print(f"Total marks: {total}")
print(f"Average marks: {average}")
```

```python
#Q7 — Highest Scoring Student
#Find highest scoring student.

students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest_student = max(students, key=students.get)
highest_score = students[highest_student]

print(f"Highest scoring student: {highest_student}")
print(f"Score: {highest_score}")
```

```python
#Q8 — Merge Dictionaries
#dict1 = {"a":1,"b":2} dict2 = {"c":3,"d":4} Merge and print result.

dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}

dict1.update(dict2)
print(dict1)
```
